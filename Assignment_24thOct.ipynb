```python
###2 Ask the user to enter a number. Handle invalid numeric input using try-except. If the number is zero or negative, raise a ValueError. If it is positive, check whether it is even or odd and display the result.

try:
    number = int(input("Enter a number"))

    if number > 0 and number%2 == 0:
        print("Number entered is even")
    elif number > 0 and number%2 == 1 :
        print("Number entered is odd")
    elif number == 0:
        raise ValueError("Number is zero")

except ValueError:
    print("Number is zero or empty or negative")
```

```python
###9
filename = input("Enter the filename to read: ")

try:
    # Open the file in read-only mode
    with open(filename, "r") as file:
        content = file.read()
        
        # Check if the file has any content
        if not content:
            print("File is empty")
        else:
            print("\n--- File Contents ---")
            print(content)
            print("---------------------")

except FileNotFoundError:
    print(f"Error: The file '{filename}' does not exist. Please check the name and try again.")
```

```python
###10 

with open("numbers.txt", "a") as file:   
    print("Please enter 5 valid, non-negative numbers.")
    i = 0
    # Continue looping until exactly 5 valid numbers are stored
    while i < 5:
        try: 
                     user_input = float(input(f"Enter number {i + 1}: "))
                     if user_input < 0:
                            raise ValueError("Invalid Number. Enter number greater than zero")
                
                     file.write(f"{user_input}\n")
                     i = i+1
                
        except ValueError as e:
       
            if "could not convert string to float" in str(e):
                print("Invalid input! Please enter a valid number (e.g., 5 or 3.14).")
            else:
                print(f"Invalid input! {e}")   

print("\nSuccess! 5 valid, non-negative numbers have been written to numbers.txt.")
```
