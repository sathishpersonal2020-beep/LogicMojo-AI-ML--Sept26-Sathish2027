```python
###questiom 39
actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]

# 1. Count the frequency of each category
category_counts = {}
for category in actual_categories:
    if category in category_counts:
        category_counts[category] += 1
    else:
        category_counts[category] = 1

# Calculate the total number of records
total_records = len(actual_categories)

# 2. Print both the counts and percentages
print("--- Category Analysis ---")
for category, count in category_counts.items():
    # Calculate percentage
    percentage = (count / total_records) * 100

 # Print the raw count and the formatted percentage
    print(f"Category {category}: Count = {count}, Percentage = {percentage:.1f}%")
```

```python
###question 40
experiment_1 = {
    "Linear Regression": 0.82,
    "Decision Tree": 0.88,
    "Random Forest": 0.94
}

experiment_2 = {
    "Decision Tree": 0.90,
    "Random Forest": 0.96,
    "KNN": 0.85
}

# 1 & 2. Merge both dictionaries, retaining the higher score for overlapping models
final_results = experiment_1.copy()  # Start with all records from experiment 1

for model, accuracy in experiment_2.items():
    if model in final_results:
        # Retain the higher score
        final_results[model] = max(final_results[model], accuracy)
    else:
        # Add the new model if it doesn't exist
        final_results[model] = accuracy

# 3. Print the final dictionary
print("--- 3. Final Merged Dictionary ---")
print(final_results)
print("----------------------------------\n")

# 4. Identify the model with the highest accuracy
best_model = None
highest_accuracy = -1.0

for model, accuracy in final_results.items():
    if accuracy > highest_accuracy:
        highest_accuracy = accuracy
        best_model = model

print(f"--- 4. Highest Accuracy Model ---")
print(f"Model: {best_model} (Accuracy: {highest_accuracy})\n")

# 5. Print all models whose final accuracy is greater than 0.90
print("--- 5. Models with Accuracy > 0.90 ---")
for model, accuracy in final_results.items():
    if accuracy > 0.90:
        print(f"- {model}: {accuracy}")
```
