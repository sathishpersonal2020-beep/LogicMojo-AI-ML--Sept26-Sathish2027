```python
###question 28
available = {"Scaling", "Encoding", "Imputation", "Feature Selection", "Normalization"}
selected = {"Scaling", "Encoding", "Imputation"}

# 1. Checking if 'selected' is a subset of 'available'
is_subset_method = selected.issubset(available)
is_subset_operator = selected <= available

# 2. Checking if 'available' is a superset of 'selected'
is_superset_method = available.issuperset(selected)
is_superset_operator = available >= selected

# Printing the results
print("--- Subset Check ---")
print("Using .issubset():", is_subset_method)
print("Using <= operator:", is_subset_operator)

print("\n--- Superset Check ---")
print("Using .issuperset():", is_superset_method)
print("Using >= operator:", is_superset_operator)
```

```python
###question 29
model_a_features = {"age", "salary", "tenure",
"credit_score", "balance"}
model_b_features = {"age", "salary", "tenure",
"department", "region"}
unique_category =model_a_features ^ model_b_features
category_both = model_a_features & model_b_features
intersection_set = model_a_features & model_b_features
union_set = model_a_features | model_b_features

# 2. Calculate sizes using len()
intersection_size = len(intersection_set)
union_size = len(union_set)

# 3. Compute Jaccard Similarity
jaccard_similarity = intersection_size / union_size
print(unique_category)
print(category_both)
print(jaccard_similarity)
```
